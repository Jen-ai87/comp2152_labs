```python
# 1. libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score

# 2. Loading the IRIS dataset
iris = load_iris()

# 3. features, labels, feature names, and label names
features = iris.data
labels = iris.target
feature_names = iris.feature_names
label_names = iris.target_names

# 4. Printing the dataset's shape
print(f"# Shape of features: {features.shape}")
print(f"# Shape of labels: {labels.shape}")

# 5. Splitting the dataset into training and testing sets
features_train, features_test, labels_train, labels_test = train_test_split(
    features, labels, test_size=0.3, random_state=42
)
print(f"# Shape of training features: {features_train.shape}")
print(f"# Shape of training labels: {labels_train.shape}")
print(f"# Shape of testing features: {features_test.shape}")
print(f"# Shape of testing labels: {labels_test.shape}")

# 6. Instantiating the DecisionTreeClassifier model
binary_mask = (labels_train == 0) | (labels_train == 1)
features_train_binary = features_train[binary_mask]
labels_train_binary = labels_train[binary_mask]

binary_mask_test = (labels_test == 0) | (labels_test == 1)
features_test_binary = features_test[binary_mask_test]
labels_test_binary = labels_test[binary_mask_test]

model = DecisionTreeClassifier()

# 7. Training the model
model.fit(features_train_binary, labels_train_binary)

# 8. Test the model based on predictions
predictions = model.predict(features_test_binary)

# 9. Print the predicted values
print("# Actual targets (first 10):", labels_test_binary[:10])
print("# Predictions (first 10):", predictions[:10])

# 10. Calculataing and printing the accuracy
accuracy = accuracy_score(labels_test_binary, predictions)
print(f"# Model accuracy: {accuracy:.2f}")

# 11. Visualizing the Decision Tree
plt.figure(figsize=(12, 8))
plot_tree(model, feature_names=feature_names, class_names=label_names[:2], filled=True)
plt.show()
```
